```python
import torch
from torchviz import make_dot
```

```python
x = torch.tensor([2.], requires_grad=True)

L = 4*x**3 - 3*x**2 + 2*x
L.backward()

print(L)
print(x.grad)
```

```python
x = torch.tensor([-1.0,2.0,4.0], requires_grad=True)

L = 3*x**2 - 2*x**2 + 5*x
L.backward(torch.ones_like(L))


print(L)
print(x.grad)
```

```python

x = torch.tensor(2.0, requires_grad=True)
y = torch.tensor(-1.0, requires_grad=True)

f = x**2*y + 3*x*y**2 + y

f.backward()

print("f:", f.item())
print("x.grad:", x.grad.item())
print("y.grad:", y.grad.item())
```

```python

x = torch.tensor(4.0, requires_grad=True)

a = 2*x + 1
b = a**3
L = torch.sqrt(b)

L.backward()

print("a:", a.item())
print("b:", b.item())
print("L:", L.item())
print("x.grad:", x.grad.item())
```

```python
pred = torch.tensor([2.0, 4.0, 6.0], requires_grad=True)
target = torch.tensor([1.0, 3.0, 5.0])

loss = ((pred - target) ** 2).mean()

loss.backward()

print("Loss:", loss.item())
print("pred.grad:", pred.grad)
```

```python
x = torch.tensor([2.0, 3.0], requires_grad=True)

a = (x ** 2).sum()
a.backward()

print("첫 번째 기울기:", x.grad)

b = (4 * x).sum()
b.backward()

print("누적된 기울기:", x.grad)

x.grad.zero_()

print("초기화된 기울기:", x.grad)
```

```python
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
y = x ** 3


gradient = torch.tensor([1.0, 0.5, -1.0])

y.backward(gradient)

print("x.grad:", x.grad)
```

```python
# 1. 입력값 생성
x = torch.tensor(2.0, requires_grad=True)

# 2. 중간 연산
a = 3 * x - 1
b = a ** 2
L = 2 * b + 5

# 3. 중간 텐서의 기울기 저장
a.retain_grad()
b.retain_grad()

# 4. 역전파
L.backward()

# 5. 결과 출력
print("L =", L)
print("x.grad =", x.grad)
print("a.grad =", a.grad)
print("b.grad =", b.grad)
```

```python

# 1. x와 z 생성
x = torch.tensor(3.0, requires_grad=True)
z = torch.tensor(2.0, requires_grad=True)

# 2. 함수 계산
f = x**2 * z + 3*z**2 + 4*x

# 3. 두 변수의 기울기를 한 번에 계산
grads = torch.autograd.grad(f, (x, z))

# 4. 결과 출력
print("f =", f)
print("기울기 =", grads)

# 5. 각 텐서의 .grad 확인
print("x.grad =", x.grad)
print("z.grad =", z.grad)
```

```python
# 1. x 생성
x = torch.tensor(2.0, requires_grad=True)

# 2. 일반적인 연산
a = x ** 2

# 3. 기울기 추적을 하지 않는 연산
with torch.no_grad():
    b = 3 * x + 2

# 4. 두 결과를 더하기
c = a + b

# 5. 역전파
c.backward()

# 6. 결과 출력
print("x.requires_grad =", x.requires_grad)
print("a.requires_grad =", a.requires_grad)
print("b.requires_grad =", b.requires_grad)
print("c.requires_grad =", c.requires_grad)

print("b.grad_fn =", b.grad_fn)
print("x.grad =", x.grad)

# 7. 별도의 텐서 생성
u = torch.tensor(3.0, requires_grad=True)

# requires_grad 변경
u.requires_grad = False

# 8. 연산
v = u ** 2

# 9. 결과 출력
print("u.requires_grad =", u.requires_grad)
print("v.requires_grad =", v.requires_grad)
print("v.grad_fn =", v.grad_fn)
```
