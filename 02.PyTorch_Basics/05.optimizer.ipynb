### Optimizer(최적화 알고리즘)
: 손실함수를 통계 계산한 기울기를 이용하여 신경망의 weight와 bias를 업데이트하는 알고리즘

- 신경망 학습의 순서
    1. 입력데이터를 모델에 전달
    2. 모델이 예측값을 계산
    3. 손실함수로 예측값과 정답의 차이를 계산
    4. loss.backward()로 각 파라미터의 Gradient를 계산
    5. optimizer.step()으로 파라미터 업데이트

- 주요 구성 요소

| 구성 요소 | 역할 |
| --- | --- |
| Loss Function | 예측값과 정답의 차이 계산 |
| Gradient | 파라미터를 조금 변경했을 때 손실이 변하는 방향과 정도 |
| Optimizer | Gradient를 이용하여 파라미터 업데이트 |
| Learning Rate | 한번 업데이트할 때 변경하는 크기 조절 |

#### 1. Gradient Decent(경사하강법)
: 손실함수가 작아지는 방향으로 파라미터를 조금씩 변경하는 방법

- Gradient가 양수이면 파라미터를 감소시키고, 음수이면 파라미터를 증가시키는 방향으로 업데이트

- 손실이 증가하는 방향을 나타내므로 Gradient의 반대방향으로 파라미터를 이동시켜 손실을 최소화

- 용어설명
    - 파라미터(Parameter) : 모델이 학습을 통해 조정하는 모든 값
    - 가중치(Weight) : 입력 데이터의 영향력을 조절하는 파라미터
    - 편향(Bias) : 입력과 별개로 출력값을 조정하는 파라미터

````python
import torch
import torch.nn as nn
import torch.optim as optim
````

````python
# 간단한 파라미터와 손실 함수
w = torch.tensor(2., requires_grad=True)
target = torch.tensor(10.)

prediction = w*3
loss = (prediction - target) ** 2

loss.backward()

print('예측값 :', prediction.item())
print('손실값 :', loss.item())
print('w의 Gradient :', w.grad.item())
print('현재 w :', w.item())
````

````python
w_old = w.item()
gradient = w.grad.item()
learning_rate = 0.1

w_new = w_old - learning_rate * gradient

print('업데이트 전 :', w_old)
print('Gradient : ', gradient)
print('Learning Rate :', learning_rate)
print('업데이트 후 :', w_new)
````

````python
w = torch.tensor(2., requires_grad=True)
target = torch.tensor(10.)

learning_rate = 0.1

for step in range(4):
    # 예측및 손실계산
    prediction = w*3
    loss = (prediction - target) ** 2

    # Gradient계산
    loss.backward()

    print(f'\n=========== {step+1} 차 학습 =============')
    print('예측값 :', prediction.item())
    print('손실값 :', loss.item())
    print('Gradient :', w.grad.item())
    print('업데이트 전 w :', w.item())

    # 파라미터 업데이트
    with torch.no_grad():
        w -= learning_rate * w.grad

    print('업데이트 후 w :', w.item())

    # 다음 학습을 위해 Gradient초기화
    w.grad.zero_()
````

#### 2. Learning Late(학습률)
: 파라미터를 한번 업데이트할 때 얼마나 크게 변경할 것인지를 결정하는 값

**Learning Rate가 너무 작은 경우**<br>
    - 파라미터가 조금씩 변경됨<br>
    - 손실이 감소하는 속도가 느릴 수 있음<br>
    - 학습에 많은 반복이 필요할 수 있음<br><br>

**Learning Rate가 너무 큰 경우**<br>
    - 파라미터가 지나치게 크게 변경이 됨<br>
    - 최적값을 지나쳐 손실이 오히려 커질수 있음<br>
    - 학습이 불안정하거나 발산할 수 있음<br>

- 하이퍼파라미터(사용자가 넣음) : 모델과 데이터에 맞게 조정해야 하는 중요한 하이퍼파라미터임.

| Optimizer | 일반적인 시작 학습률(Learning Late) |
| --- | --- |
| 경사하강법, SGD | 0.01 |
| SGD + Momentum | 0.01 ~ 0.1 |
| Adam | 0.001 |
| AdamW | 0.001 |
| RMSporp | 0.001 |

````python
# Gradient가 동일 할 때 Learning Rate에 따른 업데이트 크기 비교
w=5.0
gradient = 2.0

for lr in [0.001, 0.1, 0.5] :
    update_w = w - lr * gradient
    print(f'Learning Rate = {lr} : 업데이트 후 w = {update_w}')
````

#### 3. SGD(Stochastic Gradient Descent)
: Gradient를 이용하여 파라미터를 업데이트하는 기본적인 최적화 알고리즘

- 사용형식
torch.optim.SGD(model.parameter(), lr=0.01)

- 특징
    - 구조가 단순하고 이해하기 쉬움
    - Learning Rate설정에 영향을 많이 받음
    - momentum옵션을 사용하여 이전 업데이트 방향을 반영할 수 있음

- 기본값은 momentum을 사용하지 않은 사용하려면 별도로 지정해줘야 함

````python
model = nn.Linear(1, 1)

optimizer = optim.SGD(model.parameters(), lr=0.01)

print('Optimizer :', optimizer)
print('초기 Weight :', model.weight)
print('초기 Bias :', model.bias)
````

#### 4. optimizer.step()
: backward()로 계산한 Gradient를 이용하여 모델의 파라미터를 업데이트하는 메서드

**일반적인 순서**
```python
optimizer.zero_grad() # 이전 Gradient 제거

prediction = model(x) # 예측값 계산 
loss = loss_fn(prediction, target) # 손실 계산 

loss.backward() # gradient 계산
optimizer.step()  # 파라미터 업데이트
```

**중요!**

optimizer.step() 호출 전에 먼저 loss.backward()를 먼저 실행해야 현재 손실에 대한 Gradient를 이용할 수 있음

````python
w = torch.nn.Parameter(torch.tensor(5.))
optimizer = optim.SGD([w], lr=0.1)

target = torch.tensor(3.)

print('업데이터 전 w :', w.item())

optimizer.zero_grad()
loss = (w - target)**2
print('업데이트 전 손실 :', loss.item())
loss.backward()

print('Gradient :', w.grad.item())

optimizer.step()

print('업데이트 후 w :', w.item())
print('업데이트 후 손실 :', ((w-target)**2).item())

````

````python
x = torch.tensor([[1.0],[2.],[3.],[4.]])
target = torch.tensor([[5.],[8.],[11.],[14.]])

model = nn.Linear(1, 1)
loss_fn = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

for epoch in range(101):
    optimizer.zero_grad()

    prediction = model(x)
    loss = loss_fn(prediction, target)

    loss.backward()
    optimizer.step()

    if epoch % 20 == 0 :
        print(f'Epoch {epoch:3d} | Loss : {loss.item():.6f}')

print('학습 후 Weight :', model.weight.item())
print('학습 후 Bias :', model.bias.item())
````

#### 5. Adam(Adaptive Moment Estimation)
: 학습과정에서 Gradient의 평균적인 방향과 크기를 함께 고려하여 파라미터를 업데이트하는 최적화 알고리즘

**사용형식**
`torch.optim.Adam(model.parameters(), lr=0.001)`

- Adam의 주요 특징
    - 각 파라미터의 Gradient를 바탕으로 업데이트 크기를 적응적으로 조절함
    - Gradient의 1차 모멘트(평균)와 2차 모멘트(제곱의 평균)을 이용함
    - 일반적으로 기본 Learning Rate로 0.001을 자주 사용함
    - 다양한 신경망 학습에 널리 사용됨

- 개념적으로
    - 1차 모멘트 : Gradient의 이동평균
    - 2차 모멘트 : Gradient제곱의 이동평균
        이 두 정보를 이용하여 파라미터별 업데이트 크기를 조절함    

````python
model = nn.Linear(1,1)

optimizer = optim.Adam(model.parameters(), lr=0.001)

print('optimizer :', optimizer)
print('초기 Weight :', model.weight)
print('초기 Bias :', model.bias)
````

#### SGD와 Adam 비교
| 구분 | SGD | Adam |
| --- | --- | --- |
| 기본방식 | Gradient와 Learning Rate로 업데이트 | Gradient의 1.2차 모멘트 활용 |
| 학습률 | 사용자가 설정 | 기본적으로 적응적 업데이트 |
| 특징 | 단순하고 직관적 | 다양한 문제에서 학습이 잘 진행되는 경우가 많음 |
| 대표적인 시작값 | 0.01 등 | 0.001 등 |

````python
# 동일한 데이터로 SGD와 Adam 학습결과 비교
torch.manual_seed(42)

x = torch.tensor([[1.],[2.],[3.],[4.]])
target = 4*x-1

model_sgd = nn.Linear(1,1)
model_adam = nn.Linear(1,1)

model_adam.load_state_dict(model_sgd.state_dict())

loss_fn = nn.MSELoss()
optimizer_sgd = optim.SGD(model_sgd.parameters(), lr=0.01)
optimizer_adam = optim.Adam(model_adam.parameters(), lr=0.01)

for epoch in range(200):
    optimizer_sgd.zero_grad()
    loss_sgd = loss_fn(model_sgd(x), target)
    loss_sgd.backward()
    optimizer_sgd.step()

    optimizer_adam.zero_grad()
    loss_adam = loss_fn(model_adam(x), target)
    loss_adam.backward()
    optimizer_adam.step()

    if epoch % 50 == 0:
        print(f'Epoch [{epoch+1:3d}/200]')
        print(f'SGD Loss : {loss_sgd.item():.6f}')
        print(f'Adam Loss : {loss_adam.item():.6f}')
        print('-' * 30)

print('SGD 최종 Loss :', loss_fn(model_sgd(x), target).item())
print('Adam 최종 Loss :', loss_fn(model_adam(x), target).item())
````

#### Learning Rate에 따른 학습 결과 비교

````python
x = torch.tensor([[1.],[2.],[3.],[4.]])
target = 4*x-1

initial_model = nn.Linear(1,1)
initial_state = initial_model.state_dict()

learning_rates = [0.001, 0.03, 0.5]

for lr in learning_rates:
    model = nn.Linear(1,1)
    model.load_state_dict(initial_state)

    loss_fn = nn.MSELoss()
    optimizer = optim.SGD(model.parameters(), lr=lr)

    losses = []
    for epoch in range(30):
        optimizer.zero_grad()
        prediction = model(x)
        loss = loss_fn(prediction, target)
        loss.backward()
        optimizer.step()
        losses.append(loss.item())

    print(f'Learning Rate = {lr:<5} | 처음 Loss = {losses[0]:.4f} | 마지막 Loss = {losses[-1]:.4f}')    
````

#### Epoch와 반복학습
: Epoch는 학습 데이터 전체를 한번 학습한 단위<br><br>

예를 들어 학습 데이터가 100개이고 전체 데이터를 한번 학습했다면 1Epoch<br><br>

- 많이 학습하면 loss가 계속감소함. 그렇다고 해서 반드시 성능이 좋다라고 할 수 없음
    - 과적합(Overfitting)이 발생할 수 있음.


````python
torch.manual_seed(0)

x = torch.linspace(-2, 2, 20).reshape(-1, 1)    # (20,1)
target = 2 * x + 3

model = nn.Linear(1,1)
loss_fn == nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.05)

for epoch in range(100):
    optimizer.zero_grad()

    prediction = model(x)
    loss = loss_fn(prediction, target)

    loss.backward()
    optimizer.step()

    if epoch % 20 == 0:
        print(f'Epoch {epoch:3d} | Loss : {loss.item():.6f}')

print('최종 Weight :', model.weight.item())
print('최종 Bias :', model.bias.item())
````
