1. sigmoid 함수
: 입력값 0~1 사이 값으로 변환하는 함수 
- 특징 
- 출력 범위가 0과 1사이 
- 입력값이 0이면 출력값 0.5 
- 입력값이 커지면 출력은 1에 가까워짐 
- 입력값이 작아지면 출력은 0에 가까워짐 
- 이진분류에서 한 클래스에 속할 확률처럼 해석할 수 있는 값들을 만들 때 사용 
- 단점 
- 입력값의 절댓값이 아주 커지면 기울기 작아지는 포화현상이 나타남 
- 은닉층에 여러번 사용하면 기울기 소실 문제가 생길 수 있음 

```python
import torch
import torch.nn as nn
```

```python
x = torch.tensor([-10., -2, -1, 0, 1, 2, 10])

output =torch.sigmoid(x) 

print(output)                                                                                                                                                                                                                                                                                                                               *-
```

relu함수 
: 입력값이 0보다 작으면 0을 출력하고, 0보다 크면 입력값을 그대로 출력 

-특징 
- 계산이 단순하고 효율적임 
- 양수 영역에서는 기울기가 1이므로 기울기소실 문제가 완화 
- 일반적인 신경망의 은닉층에서 널리 사용됨 

tanh 
: 입력값이 -1 ~ 1 사이의 값으로 변환

- 특징 
- 입력값이 0이면 출력값도 0
- 양수 입력은 1에 가깝고, 음수 입력은 -1에 가까움 
- 출력 중심이 0
-단점
- 입력값의 절대값이 매우 커지면 기울기가 작아지는 포화 현상이 발생 



```python
x = torch.tensor([-10., -2, -1, 0, 1, 2, 10])

output = torch.tanh(x)

print(x)
print(output)
```

L1loss(mean absolute error)
: 예측값과 정답의 차이를 절댓값을 적용한 뒤 평균계산 

- 특징 
 - 오차에 절댓값 사용 
 

```python
prediction = torch.tensor([10.,10,10])
target = torch.tensor([9.,8,0])

mse_loss = nn.MSELoss()(prediction, target)
l1_loss = nn.L1Loss()(prediction, target)

print(mse_loss)
print(l1_loss)
```
