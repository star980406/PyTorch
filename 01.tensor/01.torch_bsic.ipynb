#### pytorch
: 인공지능과 딥러닝 모델을 학습시키기 위한 파이썬 기반의 오픈소스 머신러닝 프레임워크

- 핵심기능
 - tensor : 데이터를 다차원 배열로 저장하고 계산
 - gpu연산 : gpu를 이용한 대량의 계산을 빠르게 처리 
 - 자동미분(autograd) : 학습에 필요한 기울기를 자동 계산 
 - 최적화(optimizer) : 모델의 가중치를 수정하면서 학습 


```python
x = torch.zeros (4,2)
print(x)
print(x.dtype)

y = torch.ones(2,4)
print(y)
print(y.dtype)
```

```python
a = torch.tensor([1,2,3,4])
print (a)
print( type(a))
print(a.dtype)
print(a.shape)
print()

b = torch.tensor([1,2,3.1,4])
print(b)
print(b.dtype)
```

```python
c = torch.tensor([[1,2,3],[4,5,6]])

print(c)
print(c.shape)
print(c.ndim)
print(c.numel())
```

```python
print(torch.zeros(5))
# zeros_like (변수): 변수에 들어있는 텐서와 동일한 크기를 가지면서 모든 원소가 0인 새로운 텐서를 생성
print(torch.zeros_like(c))
print(torch.arange(3,10,2)) #range와 동일
print(torch.arange(0,1,0.1)) #소수점 가능 (range에서는 안됨)

print(torch.linspace(0,1,10)) #0~1(end포함) 10개 
```

```python
d = torch.tensor([1,2,3])
e = torch.tensor([4,5,6])
f = d + e
print(f)
```

```python
g= torch.tensor([[1,2,3],[1,2,3]])
h = torch.tensor([[4,5,6],[1,1,1]])
i = g + h
print(i)

j = g-h
print(j)

print(g*h)
print(g/h)
print(g**2)
```
