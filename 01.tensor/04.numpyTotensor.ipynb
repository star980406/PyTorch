#### numpy 배열을 tensor로 변환 
: 임베딩 벡터 처리, 데이터 전처리, 모델 입력데이터 구성 등에서 많이 사용 


```python
import numpy as np
import torch
```

```python
arr = np.array([1.0,2,3,4])
x = torch.from_numpy(arr)
print(arr)
print(x)

arr[0]= 100
print(arr)
print(x)

```

torch.tensor(array)
: 데이터를 복사, 원본과 분리된 tensor생성 

```python
arr = np.array([1.0,2,3,4])

x = torch.tensor(arr)
print(arr)
print(x)

arr[0] = 100
print(arr)
print(x)
```

-torch.as_tensor(array)
: 가능하면 메모리를 공유하고, 필요한 경우에만 데이터를 복사하는 함수
-torch.from_numpy()와 비슷하게 동작하지만, 입력 자료형이나 장치(device)등을 지정할 수 있다는 차이 

```python
arr = np.array([1.0,2,3,4])

x = torch.as_tensor(arr)

print(arr)
print(x)

arr[0]=100

print(arr)
print(x)
```

```python
arr = np.array([1,2,3,4], dtype = np.int32 )
x = torch.as_tensor(arr, dtype = torch.float32 )

print(arr)
print(x)
```

#### pytorch tensor 를 numpy 배열로 변환
: 모델이 계산한 결과를 python 의 다른 라이브러리로 전달하거나 데이터를 시각화하고 분석할 때 사용 

```python
x = torch.tensor([0.1,0.2,0.3,0.4])

arr = x.numpy()

print(x)
print(arr)
print(arr.dtype)
print(arr.shape)

x[0] = 0.7
print(arr)
print(x)
```

```python
x = torch.tensor([1.0,2,3], device = 'cuda')

# GPU -> CPU -> NUMPY 
arr = x.cpu().numpy()
```
