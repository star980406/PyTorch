문제 01
한 물류센터에서 상품별 입고량을 무작위로 생성하려고 함.

요구사항

torch.manual_seed(21)을 설정.
torch.rand()를 이용해 3행 4열의 0 이상 1 미만 실수 텐서 uniform_data 생성.
torch.randn()을 이용해 3행 4열의 표준정규분포 난수 텐서 normal_data 생성.
torch.randint()를 이용해 3행 4열의 20 이상 51 미만 정수 텐서 integer_data 생성.
세 텐서를 각각 출력.

```python
torch.manual_seed(21)
uniform_data = torch.rand(3,4)
normal_data = torch.randn(3,4)
integer_data = torch.randint(20,51,(3,4))
print(uniform_data)
print(normal_data)
print(integer_data)
```

문제 02
실험 장비에서 측정한 양수 데이터가 다음과 같음.

measurement = torch.tensor([1.0, 4.0, 9.0, 16.0])
요구사항

각 원소의 제곱근을 sqrt_data에 저장.
각 원소의 제곱을 square_data에 저장.
각 원소에 자연상수 (e)를 밑으로 하는 지수함수를 적용.
각 원소에 자연로그, 상용로그, 밑이 2인 로그를 각각 적용.
모든 결과를 구분하여 출력.

```python
measurement = torch.tensor([1.0, 4.0, 9.0, 16.0])

sqrt_data = torch.sqrt(measurement)
square_data = torch.square(measurement)
exp_data = torch.exp(measurement)
log_natural=torch.log(measurement)
log_common = torch.log10(measurement)
log_base2 = torch.log2(measurement)

print(sqrt_data)
print(square_data)
print(exp_data)
print(log_natural)
print(log_common)
print(log_base2)
```

문제 03
제품 무게를 측정한 결과가 다음과 같음. (단위: kg)

weight = torch.tensor([-2.786, -1.234, 1.236, 2.785])
요구사항

torch.abs()를 이용해 절댓값 계산.
torch.round()를 이용해 정수 단위로 반올림.
torch.round()의 decimals 옵션을 이용해 소수점 둘째 자리까지 반올림.
torch.floor(), torch.ceil(), torch.trunc()의 결과를 각각 출력.
음수에서 내림과 소수점 버림의 차이를 결과로 확인.

```python
weight = torch.tensor([-2.786, -1.234, 1.236, 2.785])

abs_data = torch.abs(weight)
round_data = torch.round(weight)
round_data2 = torch.round(weight, decimals=20)
floor_data = torch.floor(weight)
ceil_data = torch.ceil(weight)
trunc_data = torch.trunc(weight)

print( abs_data)
print( round_data)
print(round_data2 )
print(floor_data )
print(ceil_data )
print( trunc_data)
```

문제 04
세 지점에서 측정한 네 종류의 센서값이 다음과 같음.

sensor = torch.tensor([[12, 25, 18, 31],
                       [ 9, 28, 22, 17],
                       [15, 19, 35, 24]])
요구사항

전체 데이터의 최솟값과 최댓값 출력.
dim=0을 기준으로 각 열의 최댓값과 해당 인덱스 출력.
dim=1을 기준으로 각 행의 최솟값과 해당 인덱스 출력.
전체 텐서에서 최댓값과 최솟값의 위치 인덱스를 각각 출력.

```python
sensor = torch.tensor([[12, 25, 18, 31],
                       [ 9, 28, 22, 17],
                       [15, 19, 35, 24]])

print(torch.min(sensor))
print(torch.max(sensor))

max_vals,max_indices = torch.max(sensor,dim=0)
print(max_vals,max_indices)

min_vals,min_indices = torch.min(sensor,dim=1)
print(min_vals,min_indices)
```

문제 05
세 팀의 일주일 생산량을 나타내는 텐서가 다음과 같음.

production = torch.tensor([[12., 15., 18., 10.],
                           [20., 17., 16., 19.],
                           [ 8., 11., 13., 14.]])
요구사항

전체 생산량의 합계와 평균 계산.
팀별 생산량 합계를 dim=1 기준으로 계산.
팀별 평균 생산량을 dim=1 기준으로 계산하고 keepdims=True 적용.
전체 생산량의 곱, 중앙값, 표준편차, 분산 계산.
각 결과를 출력.

```python
production = torch.tensor([[12., 15., 18., 10.],
                           [20., 17., 16., 19.],
                           [ 8., 11., 13., 14.]])

total_sum = torch.sum(production)
total_mean = torch.mean(production)

team_sum = torch.sum(production,dim=1)
team_mean = torch.mean(production, dim=1, keepdims=True )

total_prod = torch.prod(production)
total_median = torch.median(production)
total_std = torch.std(production)
total_var = torch.var(production)

print(team_sum,team_mean)
print(total_prod,total_median,total_std ,total_var)
```
