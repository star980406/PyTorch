
문제 02
온도계 눈금에 사용할 두 종류의 수열 생성.

요구사항

torch.arange()를 이용해 -12부터 12까지 4 간격의 정수 생성. (12 포함)
torch.linspace()를 이용해 -12부터 12까지 양 끝을 포함한 7개의 값 생성.
두 텐서를 출력하고 값이 같은지 비교.
제약: arange()의 끝값이 포함되지 않는 점을 고려하기.

```python
t_arange = torch.arange(-12,13,4)
t_linspace = torch.linspace(-12,12,steps=7)

print(t_arange)
print(t_linspace)
```

문제 03
과일 매장의 판매 수량이 다음과 같음.

sales = torch.tensor([[6, 2, 9],
                      [4, 8, 3]], dtype=torch.float32)
요구사항

sales와 동일한 Shape과 dtype의 0 텐서를 returns에 생성.
반품 데이터에서 첫 번째 행의 두 번째 원소를 1, 두 번째 행의 세 번째 원소를 2로 변경.
net = sales - returns를 계산하고 net과 net.dtype을 출력.

```python
sales = torch.tensor([[6, 2, 9],
                      [4, 8, 3]], dtype=torch.float32)

returns = torch.zeros_like(sales)
print(returns)

returns[0,1]= 1.0
returns[1,2]=2.0

print(returns)
net = sales - returns

print(net)

print(net.dtype)
```

문제 04
제품 2종의 판매 수량과 단가가 다음과 같음.

qty = torch.tensor([[2, 3],
                    [4, 1]])
price = torch.tensor([[100, 200],
                      [300, 400]])
요구사항

같은 위치끼리 곱한 결과를 elementwise에 저장.
행렬곱 결과를 matrix_product에 저장.
두 결과를 출력하여 차이 확인.
제약: 원소별 곱은 *, 행렬곱은 @ 사용하기.

```python
qty = torch.tensor([[2, 3],
                    [4, 1]])
price = torch.tensor([[100, 200],
                      [300, 400]])

elementwise = qty*price
matrix_product = qty@price

print(elementwise)
print(matrix_product)
```

문제 08
2일 동안 측정한 3개 센서의 오전·오후 값이 다음과 같음.

measure = torch.tensor([[[11, 12], [21, 22], [31, 32]],

Shape (2, 3, 2)는 (날짜, 센서, 시간대) 를 의미함.

문제 09
세 명의 선수 기록을 나타내는 행렬이 다음과 같음.

문제 12
창고의 온도 측정값(°C)이 다음과 같음.

temp = torch.tensor([[-6,  4, 18, 31],
                     [ 0, 12, 27, 39],
                     [ 7, 22, 35, -3]])
요구사항

temp에서 0 미만이거나 30 초과인 값만 1차원 텐서로 추출.
원본을 보존하고 safe_temp라는 복사본 생성.
safe_temp에서 0 미만인 값을 0으로 변경.
변경 전후 텐서 출력.
제약: 조건 결합에는 |, 복사에는 clone() 사용하기.

```python
temp = torch.tensor([[-6,  4, 18, 31],
                     [ 0, 12, 27, 39],
                     [ 7, 22, 35, -3]])

mask = (temp < 0) | (temp > 30)
extreme_temps = temp[mask]
print(extreme_temps)

safe_temp = temp.clone()
safe_temp[safe_temp<0]=0

print(temp)
print(safe_temp)
```
